## 1. Programs that make decisions

So far, our programs have run every line, top to bottom. Real programs need to **choose**: *if it's raining, bring an umbrella.*

Remember the logic gates notebook? Every comparison gives a `True` or `False`. A **conditional** uses that answer to decide what to run.

````python
age = 16

if age >= 13:
    print("You can make a teen account.")
````

Read it like English: *if age is at least 13, then print the message.*

The pattern has three important parts:

```python
if condition:
    code that runs only when the condition is True
```

1. The **condition** is anything that gives `True` or `False`
2. The line ends with a **colon** `:`
3. The code that belongs to the `if` is **indented** (pushed in by 4 spaces)

**Try it:** change `age` to `10` and run the cell. Nothing prints, because the condition is `False`, so Python skips the indented line.

## 2. Indentation matters

In Python, the **indentation** shows which lines belong to the `if`. Lines that are *not* indented always run.

````python
temperature = 35

if temperature < 40:
    print("It's cold.")
    print("Wear a coat.")

print("Have a nice day!")
````

The first two `print` lines are indented, so they only run when it's cold. The last line is not indented, so it runs **no matter what**.

**Try it:** change `temperature` to `75` and run the cell. Only the last line prints.

## 3. `else`: the other choice

`else` says what to do when the condition is `False`. Exactly **one** of the two blocks runs.

````python
score = 55

if score >= 60:
    print("You passed!")
else:
    print("You did not pass yet.")
````

Notice that `else` has no condition of its own, just a colon. It means "in every other case."

**Try it:** change `score` to `60` and run it again. Is 60 a pass? Why? (Check the `>=`.)

## 4. `elif`: more than two choices

What if there are more than two outcomes, like letter grades? `elif` (short for "else if") adds more conditions. Python checks them **in order** and runs the **first** one that is `True`, then skips the rest.

````python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: below C")
````

Why does `85` print `B` and not also `C`? Because once `score >= 80` was `True`, Python skipped everything after it.

**Order matters.** Look at what happens if we check the lowest grade first:

````python
score = 95

if score >= 70:
    print("Grade: C")
elif score >= 80:
    print("Grade: B")
elif score >= 90:
    print("Grade: A")
````

A score of 95 gets a C! The first condition (`score >= 70`) was already `True`, so Python never looked at the others. When using `elif`, put the most specific conditions first.

**Try it:** can you fix the order so 95 prints `A`?

## 5. Comparison operators

Conditions usually come from comparisons:

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |

**Warning:** `=` and `==` are different!
- `=` **assigns** a value to a variable (`age = 16`)
- `==` **asks a question**: are these equal? (`age == 16`)

````python
password = "python123"

if password == "python123":
    print("Access granted")
else:
    print("Access denied")
````

Strings can be compared too. Python is case-sensitive, so `"Python123"` is **not** equal to `"python123"`. Try changing the password in the cell above.

You can also check whether a piece of text appears inside another with `in`:

````python
word = "banana"

if "a" in word:
    print("The word contains an a.")

if "z" in word:
    print("The word contains a z.")
else:
    print("No z in this word.")
````

## 6. Combining conditions: `and`, `or`, `not`

These are the same logic gates from the earlier notebook, now used to make decisions.

- `and`: **both** conditions must be `True`
- `or`: **at least one** must be `True`
- `not`: flips `True` and `False`

````python
age = 15
has_ticket = True

if age >= 13 and has_ticket:
    print("Enjoy the movie!")
else:
    print("Sorry, you can't come in.")
````

````python
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a school day.")
````

````python
is_raining = False

if not is_raining:
    print("No umbrella needed.")
````

A common mistake is writing `if day == "Saturday" or "Sunday":`. Python doesn't read that like English, so always repeat the full comparison on both sides of `or`.

## 7. Nested conditionals

You can put an `if` **inside** another `if`. The inner one only gets checked when the outer one is `True`. Each level is indented further.

````python
has_account = True
password_correct = False

if has_account:
    if password_correct:
        print("Welcome back!")
    else:
        print("Wrong password.")
else:
    print("Please create an account.")
````

Nested decisions can often be written more simply with `and`. Try rewriting the example above, so it only prints "Welcome back!" when both are `True`.

## 8. Common mistakes

Here are the errors beginners hit most often. These are shown as text so they don't break the notebook, but you'll fix real ones in the practice section.

**1. Missing colon**
```python
if age >= 13
    print("Teen")
```
Python says `SyntaxError: expected ':'`

**2. Using `=` instead of `==`**
```python
if age = 13:
    print("Thirteen")
```
Python says `SyntaxError`. Use `==` to compare.

**3. Forgetting to indent**
```python
if age >= 13:
print("Teen")
```
Python says `IndentationError: expected an indented block`

**4. Indenting inconsistently**
Every line in the same block must be indented the same amount. Use 4 spaces each time.

- **Indentation** (4 spaces) shows which code belongs to each branch, and every `if`/`elif`/`else` line ends with a **colon**
