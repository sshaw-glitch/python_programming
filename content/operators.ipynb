## 1. Arithmetic operators

An **operator** is a symbol that tells Python to do something with values. You already know the basic math ones. Python has a few extra.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide | `7 / 2` | `3.5` |
| `//` | floor divide (whole-number part) | `7 // 2` | `3` |
| `%` | modulo (remainder) | `7 % 2` | `1` |
| `**` | power (exponent) | `7 ** 2` | `49` |

````python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
````

Operators work on variables too:

````python
apples = 12
friends = 5

print(apples + friends)
print(apples * friends)
````

## 2. Three kinds of division

Dividing 17 cookies among 5 people gives three useful answers.

````python
cookies = 17
people = 5

print(cookies / people)     # exact answer, with a decimal
print(cookies // people)    # how many whole cookies each person gets
print(cookies % people)     # how many cookies are left over
````

- `/` gives the exact answer as a decimal (a `float`)
- `//` gives only the **whole-number part**, dropping the remainder
- `%` gives only the **remainder**

Notice that `//` and `%` go together: 17 cookies = 3 each × 5 people + 2 left over.

Notice also that `/` **always** gives a float, even when it divides evenly:

````python
print(6 / 3)
print(type(6 / 3))
print(6 // 3)
print(type(6 // 3))
````

## 3. What is the remainder operator good for?

`%` seems odd at first, but programmers use it all the time.

**Even or odd?** A number is even if dividing by 2 leaves remainder 0.

````python
print(8 % 2)    # 0 means even
print(9 % 2)    # 1 means odd
````

**Clocks and cycles.** If it's 10 o'clock, what time is it 5 hours later? The answer wraps around after 12, which is exactly what `%` does.

````python
hour = 10
hours_later = 5

print((hour + hours_later) % 12)
````

It prints `3`, because 15 wraps around to 3 on a 12-hour clock.

## 4. Order of operations

Python follows the same order you learned in math:

1. Parentheses `( )`
2. Exponents `**`
3. Multiplication, division, `//`, and `%` (left to right)
4. Addition and subtraction (left to right)

````python
print(7 + 3 * 2)
print((7 + 3) * 2)
````

The first line does `3 * 2` first, giving `13`. The second does the parentheses first, giving `20`. **When in doubt, add parentheses.** They make your code easier to read too.

## 5. Shortcut operators

Updating a variable is so common that Python has shortcuts.

````python
score = 10

score += 5     # same as score = score + 5
print(score)

score -= 2     # same as score = score - 2
print(score)

score *= 2     # same as score = score * 2
print(score)
````

There are also `/=`, `//=`, `%=`, and `**=`, which work the same way.

## 6. Operators on strings

Some operators also work on text (strings), but they mean something different.

````python
first = "Hello"
second = "World"

print(first + " " + second)    # + joins strings together
print(first * 3)               # * repeats a string
````

Other operators, like `-` and `/`, don't work on strings. And you **can't mix** text and numbers with `+`. **This next cell produces an error on purpose.** Run it and read the message.

````python
age = 15
print("I am " + age + " years old")
````

Python can't add text and a number. There are two fixes:

````python
age = 15

print("I am " + str(age) + " years old")   # str() turns a number into text
print(f"I am {age} years old")             # an f-string handles it for you
````

## 7. Length, indexing, and slicing

`len()` tells you how many characters a string has.

Each character has a position number called an **index**. **Indexes start at 0**, not 1.

```
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

Negative numbers count from the end, so `-1` is the last character.

````python
word = "Python"

print(len(word))
print(word[0])
print(word[1])
print(word[-1])
````

A **slice** gets a piece of a string: `word[start:stop]` includes `start` but stops *before* `stop`.

````python
word = "Python"

print(word[0:3])    # characters 0, 1, 2
print(word[:2])     # from the beginning up to (not including) 2
print(word[2:])     # from 2 to the end
````

## 8. A few string methods

A **method** is an action attached to a value, written with a dot.

````python
name = "ada lovelace"

print(name.upper())
print(name.title())
print(name.replace("a", "A"))
````

Methods don't change the original variable. They give you a **new** string back.

````python
print(name)
````
