## 1. One switch, one light

Picture a battery, a switch, and a light bulb in a loop.

- Switch **closed** (on) → electricity flows → light is **ON**
- Switch **open** (off) → no electricity → light is **OFF**

That's the simplest circuit, and it already does something useful: the light's state depends on the switch's state.

Computers describe "on" and "off" with two values:

| Electricity | Python | Binary |
|-------------|--------|--------|
| on | `True` | `1` |
| off | `False` | `0` |

````python
switch = True
print("Is the light on?", switch)
````

Change `True` to `False` and run it again. `True` and `False` are called **booleans**, a data type named after the mathematician George Boole, who studied logic.

## 2. Setup: run this cell first

The cell below defines some helper tools we'll use to build "logic gates" and print tables. **You don't need to understand this code yet**, just run it once. (If you ever see a `NameError` later, come back and run this cell.)

````python
def not_gate(a):
    return not a

def and_gate(a, b):
    return a and b

def or_gate(a, b):
    return a or b

def xor_gate(a, b):
    return a != b

def truth_table_1(gate):
    print("A       | Output")
    print("--------------------")
    for a in (False, True):
        print(f"{str(a):<7} | {gate(a)}")

def truth_table(gate):
    print("A       B       | Output")
    print("---------------------------")
    for a in (False, True):
        for b in (False, True):
            print(f"{str(a):<7} {str(b):<7} | {gate(a, b)}")

print("Ready!")
````

## 3. Logic gates

A **logic gate** is a small circuit that takes one or more inputs (on/off) and produces an output (on/off) following a simple rule. Computer chips are made of huge numbers of them, built from tiny switches called **transistors**. A transistor is a switch controlled by electricity instead of a finger.

A **truth table** lists every possible input and the output the gate gives.

### NOT gate: flips the input

Input on → output off. Input off → output on.

````python
print(not True)
print(not False)
````

````python
truth_table_1(not_gate)
````

### AND gate: both must be on

Imagine **two switches in a row** (in series) on the same wire. The light only turns on if switch A **and** switch B are both closed.

````python
switch_a = True
switch_b = False

print("Light is on:", switch_a and switch_b)
````

````python
truth_table(and_gate)
````

**Try it:** change `switch_b` to `True` in the cell above and run it again. Only one combination makes an AND gate output `True`. Which one?

### OR gate: at least one must be on

Now imagine **two switches side by side** (in parallel). Electricity has two paths, so the light turns on if switch A **or** switch B (or both) is closed.

````python
switch_a = True
switch_b = False

print("Light is on:", switch_a or switch_b)
````

````python
truth_table(or_gate)
````

### XOR gate: exactly one must be on

**XOR** ("exclusive or") is `True` when the inputs are **different**. In Python we can write it with `!=` (not equal).

````python
truth_table(xor_gate)
````

## 4. Combining gates

Gates can be connected so the output of one becomes the input of another. In Python we do this by combining `and`, `or`, and `not` in one expression.

**Example: a house alarm.** It should sound if a door **or** a window is open, **and** the system is armed.

````python
door_open = True
window_open = False
armed = True

alarm_sounds = (door_open or window_open) and armed
print("Alarm sounds:", alarm_sounds)
````

The parentheses work like in math: the `or` is worked out first, then its result is combined with `and`.

**Try it:** set `armed = False` and run the cell again. Then try `door_open = False`.

Big chips are just this idea repeated: gates feeding into gates, feeding into gates. Each layer builds something more complex on top of the one beneath. This is called **abstraction**.

## 5. Adding numbers with logic

Here's the surprising part: a computer can **add** using only logic gates.

Add two single bits (each `0` or `1`):

| A | B | Sum | Result in binary |
|---|---|-----|------------------|
| 0 | 0 | 0 | `00` |
| 0 | 1 | 1 | `01` |
| 1 | 0 | 1 | `01` |
| 1 | 1 | 2 | `10` |

The result has two parts: a **sum bit** (right digit) and a **carry bit** (left digit). Look at the pattern:

- The **sum bit** is 1 when the inputs are *different*, which is an **XOR** gate
- The **carry bit** is 1 only when *both* inputs are 1, which is an **AND** gate

This little circuit is called a **half adder**.

````python
a = True    # this stands for the bit 1
b = True    # this stands for the bit 1

sum_bit = a != b        # XOR
carry_bit = a and b     # AND

print("Sum bit:", sum_bit)
print("Carry bit:", carry_bit)
print("So 1 + 1 gives binary:", int(carry_bit), int(sum_bit))
````

`int(True)` turns `True` into `1`, and `int(False)` into `0`. The output `1 0` is binary for the number 2.

**Try it:** change `a` and `b` to try all four combinations and compare with the table above. Real computers chain many of these adders together to add big numbers.

## 6. Where does memory come from?

Gates by themselves just react to their *current* inputs. So how does a computer *remember* something?

By connecting a gate's output **back into its own input**, so the circuit holds its value even after the input goes away. One such circuit stores one bit. Millions of them together form the computer's memory.

## 7. Python asks yes/no questions too

Python constantly produces booleans when you **compare** values:

| Symbol | Meaning |
|--------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` `>` | less than, greater than |
| `<=` `>=` | less than or equal, greater than or equal |

````python
age = 15
print(age >= 13)
print(age == 16)
print(age >= 13 and age < 18)
````

Every `if` decision a program makes comes down to a `True` or `False` like these, which a chip ultimately computes with gates.
