## 1. What is a variable?

A **variable** is a name that refers to a value stored in the computer's memory. You can think of it as a label on a box: the label is the name, and the box holds the value.

We create a variable using the **assignment operator**, `=`.

````python
age = 16
print(age)
````

Here `age` is the variable name and `16` is the value. In programming, `=` does **not** mean "equals" like in math. It means: *take the value on the right and store it under the name on the left.*

You can make as many variables as you need:

````python
player_name = "Ada"
score = 0
level = 1

print(player_name)
print(score)
print(level)
````

## 2. Variables can change

The name stays the same, but the value it holds can be replaced. That's why they're called *variables*.

````python
score = 0
print("Score:", score)

score = 10
print("Score:", score)

score = score + 5
print("Score:", score)
````

Look closely at `score = score + 5`. The computer works out the **right side first** (`10 + 5 = 15`), then stores the result back in `score`.

**Try it:** change the `5` to another number and run the cell again.

## 3. Data types

Every value has a **type**. Python figures out the type from the value you give it.

| Type | Meaning | Example |
|------|---------|---------|
| `int` | whole number | `42` |
| `float` | number with a decimal point | `3.14` |
| `str` | text (a "string"), written in quotes | `"Hello"` |
| `bool` | True or False | `True` |

````python
whole_number = 42
decimal = 3.14
text = "Hello, world!"
is_ready = True

print(type(whole_number))
print(type(decimal))
print(type(text))
print(type(is_ready))
````

Types matter because they control what you can do with a value. Numbers can be added, and strings can be joined together:

````python
print(3 + 4)          # adding numbers
print("3" + "4")      # joining strings
print("Ha" * 3)       # repeating a string
````

Notice that `3 + 4` gives `7` but `"3" + "4"` gives `"34"`. The quotes make `"3"` text, not a number.

## 4. Naming variables

Rules Python enforces:
- Names can contain letters, digits, and underscores (`_`)
- Names **cannot start with a digit**
- Names **cannot contain spaces**
- Names are **case-sensitive** (`score` and `Score` are different variables)
- Names cannot be Python keywords like `if`, `for`, `class`, or `True`

Good habits:
- Use **descriptive** names: `total_price` is better than `x`
- Use `snake_case` (lowercase words separated by underscores)

````python
my_score = 10
My_Score = 20

print(my_score)
print(My_Score)
````

Two different variables, because the capital letters make the names different.

## 5. Using variables in calculations

Variables can be used anywhere you'd use a value.

````python
width = 8
height = 5

area = width * height
print("Area:", area)
````

### Putting variables inside text: f-strings

An **f-string** lets you insert variables into text. Put an `f` before the opening quote and wrap variables in curly braces `{ }`.

````python
name = "Ada"
age = 16

print(f"{name} is {age} years old.")
print(f"Next year, {name} will be {age + 1}.")
````

## 6. Common error: using a variable that doesn't exist

If you use a name before you've created it, Python raises a `NameError`. **This next cell produces an error on purpose.** Run it and read the message.

````python
print(favorite_color)
````

The error says `favorite_color` is not defined. Fix it by creating the variable first, then printing it:

````python
favorite_color = "blue"
print(favorite_color)
````
